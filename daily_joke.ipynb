```python
import requests
import json
import pandas as pd
```

```python
def extract_joke_from_API() -> pd.DataFrame:
  '''Requests from joke API to extract a random joke,
      then returns a pandas Data Frame with the returned data'''
  try:
    url = "https://official-joke-api.appspot.com/jokes/random"
    joke_data = requests.get(url)
  except:
    raise Exception("Joke API not available")
  else:
    return pd.DataFrame(joke_data.json(), index = [0])
```
